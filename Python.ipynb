```python
import pandas as pd
```

```python
df = pd.read_excel("Sales_Data.xlsx", header=2)

# set first row as header
df.columns = df.iloc[0]

# remove that row
df = df[1:]

df.head()
```

```python
df = df.loc[:, ~df.columns.isna()]
df = df.loc[:, df.columns != True]
```

```python
df = df.dropna(how='all')
```

```python
df = df.reset_index(drop=True)
```

```python
df['Order Date'] = pd.to_datetime(df['Order Date'], dayfirst=True, errors='coerce')
```

```python
df = df.dropna(how='all')
df = df.drop_duplicates()
df = df.reset_index(drop=True)

df['Order Date'] = pd.to_datetime(df['Order Date'], dayfirst=True, errors='coerce')

df[['Car Brand', 'Car Model']] = df['Product'].str.split('/', expand=True)
```

```python
df['No.of Cars'] = pd.to_numeric(df['No.of Cars'], errors='coerce')
df['Price per car'] = pd.to_numeric(df['Price per car'], errors='coerce')
df['Total Amount'] = pd.to_numeric(df['Total Amount'], errors='coerce')
```

```python
total_sales = df['Total Amount'].sum()
avg_price = df['Price per car'].mean()
total_cars = df['No.of Cars'].sum()

print("Total Sales:", total_sales)
print("Avg Price:", avg_price)
print("Total Cars:", total_cars)
```

```python
# sales by region
region_sales = df.groupby('Region Code')['Total Amount'].sum().reset_index()

# sales by month
month_sales = df.groupby('Month')['Total Amount'].sum().reset_index()
```

```python
df.to_excel("Cleaned_Sales_Data.xlsx", index=False)
```
